```python
def to_symbol (digit):
        if digit < 10:
                return str(digit)
        return chr(ord('A') + digit - 10)

class Number:
        def __init__ (self, number, base):
                self.digits = []
                self.number = number
                self.base = base
                while (number > 0):
                        self.digits.append(number % base)
                        number = number // base

        def write_transformation (self): 
                number = self.number
                while (number > 0):
                        print (number, "\t\\", self.base, "\t=", number // self.base, "\tR", number % self.base)
                        number = number // self.base

        def return_number(self):
                t = ""
                for i in range(len(self.digits)):
                        t = to_symbol(self.digits[i]) + t
                return t


def print_addition (a, b):
        a_str = a.return_number()
        b_str = b.return_number()
        c = Number(a.number + b.number, b.base)
        c_str = c.return_number()

        max_len = 16

        print("\t", " " * (max_len - len(a_str)), a_str)
        print(" +\t", " " * (max_len - len(b_str)), b_str)
        print("--------------------------")
        print("\t", " " * (max_len - len(c_str)), c_str)
        return c
```

```python
number_a = 2032
number_b = 3040

base = 2


a = Number(number_a, base)
b = Number(number_b, base)

a.write_transformation()
print()
b.write_transformation()
print()

print_addition(a, b)
```

```python
import random

width = 15
height = 20

possible_bases = [i for i in range(2, 17)]

bases = possible_bases

basic_numbers = [random.randint(10,1000) for i in range(height)]
#basic_numbers = [45, 142, 55, 124, 164]


table = [Number(basic_numbers[i // width], bases[i % width]) for i in range(width * height)]

def print_table(table, print_whole):
        indexes = []
        spaces = [3 for i in range(width)]
        for l in range(height * width):
                i = l //width
                index = l % width
                spaces[index] = max(spaces[index], len(table[i * width + index].return_number())+1)
                
        for l in range(height):
                index = l %width
                indexes.append(index)


        print("| ", end = "")
        for i in range(width):
                print((spaces[i] - len(str(bases[i]))) * " " + str(bases[i]), "| ", end = "")
        print()

        for i in range(height):
                print("| ", end = "")
                for j in range(width):
                        if j == indexes[i] or print_whole: 
                                s = table[i*width + j].return_number()
                                print((spaces[j] - len(s)) * " " + s, "| ", end = "")
                        else:
                                print((spaces[j] + 1) * " ", end = "| ")

                print()

print_table(table, False)
print()
print_table(table, True)
```
